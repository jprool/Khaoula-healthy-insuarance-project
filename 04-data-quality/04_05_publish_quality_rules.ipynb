# Publish Data Quality Rules

## Purpose

In this notebook, I publish the approved data-quality contracts into the
central Unity Catalog quality-rules repository.

The profiling notebooks are used to investigate and approve rules.

This notebook is responsible for publishing those approved rules into:

`health_insurance.governance.quality_rules`

The Lakeflow pipeline will later retrieve active rules from this table and
apply them as expectations during Bronze-to-Silver processing.

### Publishing approach

I use an idempotent MERGE operation so that:

- new rules are inserted
- existing rules are updated
- rerunning the notebook does not create duplicates
- rule severity can be changed safely
- rules can be enabled or disabled independently
- rule metadata remains centrally governed
